# Import Dataset dari Hugging Face

Dataset **`sewa-rural-care/anemia-survey-dataset`** adalah *gated dataset*.
Sebelum proses di bawah bisa jalan, lakukan **satu kali setup akses**:

1. Login akun di [huggingface.co](https://huggingface.co) dan minta akses di
   https://huggingface.co/datasets/sewa-rural-care/anemia-survey-dataset
   (isi form Data Access Agreement & Submit).
2. Buat token **Read** di https://huggingface.co/settings/tokens
3. Autentikasi terminal: `hf auth login` (paste token).

Kalau skip langkah ini akan muncul error `401 Unauthorized` / `GatedRepoError`.

```python
from datasets import load_dataset

ds = load_dataset("sewa-rural-care/anemia-survey-dataset", split="train", token=True)
print(ds)
```

### Inspeksi data
Baca ke pandas, lihat kolom & distribusi label anemia.

```python
import pandas as pd

df = ds.to_pandas()
print(df.columns.tolist())
print(df["hemoglobin_category"].value_counts())
```

### Opsi streaming (disarankan untuk eksplorasi)
Dataset total ~836 GB. Streaming menghindari download penuh:

```python
ds_stream = load_dataset("sewa-rural-care/anemia-survey-dataset", split="train", streaming=True, token=True)
row = next(iter(ds_stream))
print(row["patient_uuid"], row["hemoglobin_category"])
```
